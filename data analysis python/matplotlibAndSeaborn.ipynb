```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5,6,7,8]
y=[3,4,5,33,43,66,76,82]



plt.plot(x,y, color='red', linestyle='--', marker='o', linewidth=3, markersize=8)
plt.xlabel('dd')
plt.grid(True)

```

```python
x=[1,2,3,4,5,6,7,8]
y1=[3,4,5,33,43,66,76,82]
y2=[3,4,5,3,4,6,76,82]

plt.figure(figsize=(10,6))

plt.subplot(1,2,1)
plt.plot(x,y1, color='red', linestyle='--', marker='o', linewidth=3, markersize=8)
plt.xlabel('dd')
plt.subplot(1,2,2)
plt.plot(x,y2, color='red', linestyle='--', marker='o', linewidth=3, markersize=8)
plt.xlabel('ddkjfbj')
plt.grid(True)
```

```python
import seaborn as sns

tips=sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
```

```python
sns.lineplot(x='size',y='total_bill', data=tips)
```

```python
sns.barplot(x='day',y='tip', data=tips)
```

```python
sns.boxplot(x='day',y='total_bill', data=tips)
```

```python
sns.violinplot(x='day',y='total_bill', data=tips)
```

```python
sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
sns.kdeplot(tips['total_bill'], fill=True)
```

```python
sns.pairplot(tips)
```

```python
corr=tips[['total_bill','tip','size']].corr()
corr
sns.heatmap(corr,annot=True)
```
