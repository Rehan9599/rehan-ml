```python
import sqlite3
```

```python
connection=sqlite3.connect('ex.db')
connection
```

```python
cursor=connection.cursor()
cursor
```

```python
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key, 
    name Text Not Null, 
    age Integer, 
    department text 
    )
''')

connection.commit()
```

```python
cursor.execute('''
    select * from employees            
''')
```

```python
cursor.execute('''
insert into employees(id,name,age,department)
  values(2,'rren',22,'cse')
''')
cursor.execute('''
insert into employees(id,name,age,department)
  values(3,'rten',12,'cse')
''')

connection.commit()
```
