# minimum , Q1 , median , Q3 , maximum

```python
import numpy as np

list_marks=[0,-40,39,38,45,50,60,65,56,45,38,67,88,77,55,98,99,92]
min,q1,med,q3,max=np.quantile(list_marks,[0.0,0.25,0.5,0.75,1.0])
min,q1,med,q3,max
```

```python
IQR=q3-q1
IQR
```

```python
lower_fence=q1-1.5*IQR
higher_fence=q3+1.5*IQR

lower_fence,higher_fence
```

```python
import seaborn as sns

sns.boxplot(list_marks)
```
